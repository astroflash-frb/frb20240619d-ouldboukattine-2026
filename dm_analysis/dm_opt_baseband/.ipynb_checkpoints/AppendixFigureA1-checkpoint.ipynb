```python
# Import packages
import numpy as np
import your
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib import colors
from scipy.optimize import curve_fit, leastsq
from tqdm import tqdm
# Import FRB handling functions
import basic_funcs
import super_dedisperse
```

```python
def gaus(x,a,x0,sigma,d):
    return a*np.exp(-(x-x0)**2/(2*sigma**2))+d

def sn_vs_dm_3peaks(peaks1,peaks2,peaks3,titulo,myrange,filename):
    """
    Given 3 sets of S/N vs DM data, fit Gaussians to each as well as the combined data
    
    Input:
    - peaks1: S/N vs DM of microshot 1
    - peaks2: S/N vs DM of microshot 2
    - peaks3: S/N vs DM of microshot 3
    - titulo: Identifier to print in top right corner
    - myrange: np.arange with DMs used
    - filename: save file name
    
    """
    
    # Set up figure
    fig,ax=plt.subplots(figsize=(7,5))
    
    # Fit each microshot with a Gaussian
    popt1,pcov1=curve_fit(gaus,myrange,peaks1,p0=[500,219.35,.1,50],maxfev=10000)
    popt2,pcov2=curve_fit(gaus,myrange,peaks2,p0=[500,219.35,.1,50],maxfev=10000)
    popt3,pcov3=curve_fit(gaus,myrange,peaks3,p0=[500,219.35,.1,50],maxfev=10000)
    # Plot the S/N vs DM
    plt.plot(myrange,peaks3,'k.')
    plt.plot(myrange,peaks2,'k.')
    plt.plot(myrange,peaks1,'k.')
    # Plot the Gaussian fits
    plt.plot(myrange,gaus(myrange,*popt1),color='turquoise',lw=3.5,alpha=0.65,label='Individual microshot')
    plt.plot(myrange,gaus(myrange,*popt2),color='turquoise',lw=3.5,alpha=0.65)
    plt.plot(myrange,gaus(myrange,*popt3),color='turquoise',lw=3.5,alpha=0.65)

    
    # Combine the peaks
    c_peaks=np.sum([peaks1,peaks2,peaks3],axis=0)
    poptc,pcovc=curve_fit(gaus,myrange,c_peaks,p0=[500,219.37,.1,100],maxfev=10000)
    plt.plot(myrange,c_peaks,'.',color='k')
    plt.plot(myrange,gaus(myrange,*poptc),color='darkviolet',lw=3.5,alpha=0.75,label='Combined')
    # Plot the peak of the fit and error
    plt.axvline(poptc[1],color='darkorange',lw=2)
    plt.axvspan(xmin=poptc[1]-3*poptc[2]/np.sqrt(len(myrange)),xmax=poptc[1]+3*poptc[2]/np.sqrt(len(myrange)),ymin=0,ymax=1,color='darkorange',alpha=0.25)

    # Style stuff
    for axis in ['top', 'bottom', 'left', 'right']:
        ax.spines[axis].set_linewidth(1.5)
        ax.tick_params(axis='both', which='major', labelsize=12)
    plt.text(s=titulo,x=219.45,y=0.9*max(c_peaks),size=14)
    plt.xlabel('DM (pc/cm$^3$)',size=14)
    plt.ylabel('S/N',size=14)
    plt.xlim([219.2,219.52])
 #   plt.ylim([200,3000])
    plt.legend(loc=2,fontsize=13)
    
    plt.savefig(filename+'.pdf',dpi=300)
    print('DM: {} pc/cc +- {} (3sigma)'.format(np.round(poptc[1],3),np.round(3*poptc[2]/np.sqrt(len(myrange)),3)))
    
    return
```

```python
def prep_burst(y,dm,n1,n2):
    """y - your object"""
    dm=dm
    tsamp=16e-6
    
    #extract burst
    nstart = int( (100e-3/tsamp)-(10e-3/tsamp) ) #Burst at 100ms - 10ms window
    nread = y.your_header.nspectra - nstart 
    d = y.get_data(nstart,nread,npoln=4)
    
    #get coherence products
    aa = d[:,0,:].T #time increasing, freq decreasing
    bb = d[:,1,:].T
    cr = d[:,2,:].T
    ci = d[:,3,:].T
    
    #make stokesi 
    stokes_i = aa + bb
    #dedisperse
    stokes_i = super_dedisperse.super_dedisperse(stokes_i,dm,freqs,tsamp,10) 
    #normalize
    norm_v1=n1 
    norm_v2=n2 
    stokes_i = basic_funcs.normalize(stokes_i,stokes_i[:,norm_v1:norm_v2])

    #make profile
    prof = np.mean(stokes_i,axis=0)
    offprof = prof[n1:n2]
    offprof_std = np.std(offprof)
    offprof_mean = np.mean(offprof)   
    stokes_i_normprof = (prof - offprof_mean) / offprof_std
    stokes_i_normprof = stokes_i_normprof[0:int(60e-3/tsamp)] #hardcoded 60ms
    
    #crop
    stokes_i = stokes_i[:,0:int(60e-3/tsamp)] #hardcoded 60ms 

    return stokes_i, stokes_i_normprof


```

```python
freqs=np.linspace(1738,1230,128)
```

```python
#load in bursts into YOUR objects
y1=your.Your('b_59881.fil')
y2=your.Your('b_59884.fil')
y3=your.Your('b_59899.fil')

#get dynamic spectra and profiles
d1,p1 = prep_burst(y1,219.356,9000,12500)
d2,p2 = prep_burst(y2,219.377,12500,15000)
d3,p3 = prep_burst(y3,219.8,9000,11500)
```

```python
#create empty lists for max vals of shots
pk1a=[]
pk1b=[]
pk1c=[]

pk2a=[]
pk2b=[]
pk2c=[]

#dedisperse to a range of DMs, find max of the peak of microshots
for i in tqdm(np.arange(219.2,219.55,.001)):
    d1,p1 = prep_burst(y1,i,9000,12500)
    d2,p2 = prep_burst(y2,i,12500,15000)

    #range manually selected to make sure the microshot is there at min and max dm being searched
    pk1a.append(max(p1[1040:1075]))
    pk1b.append(max(p1[1100:1130]))
    pk1c.append(max(p1[1135:1160]))
    
    pk2a.append(max(p2[700:740]))
    pk2b.append(max(p2[750:790]))
    pk2c.append(max(p2[825:875]))  
```

```python
sn_vs_dm_3peaks(pk1a,pk1b,pk1c,'B1',np.arange(219.2,219.55,.001),'AppendixFigureA1_B1')
```
